### Exercice 2 ⭐ – Taux d'occupation d'un disque

Un disque de `512` Go contient `389` Go de données. Calculez :
- `free_gb` : l'espace libre en Go ;
- `usage_percent` : le pourcentage d'occupation (de 0 à 100).

````python
total_gb = 512
used_gb = 389
free_gb = total_gb - used_gb
usage_percent = used_gb / total_gb * 100
print("Libre :", free_gb, "Go – occupation :", usage_percent, "%")
````

<details><summary>💡 Indice</summary>Pourcentage = partie / total × 100. L'opérateur <code>/</code> donne toujours un résultat décimal.</details>

````python
# ✅ Vérification
assert free_gb == 123, "free_gb devrait valoir 123"
assert abs(usage_percent - 75.98) < 0.01, "usage_percent devrait valoir environ 75.98"
print("✅ Exercice 2 réussi")
````

### Exercice 3 ⭐⭐ – Budget d'achat de postes

Le service informatique doit acheter `12` ordinateurs portables à `1450` dinars HT l'unité. La TVA est de 19 %.

````python
laptop_count = 12
unit_price_ht = 1450
vat_rate = 0.19

total_ht = laptop_count * unit_price_ht
vat_amount = total_ht * vat_rate
total_ttc = total_ht + vat_amount
print("Total HT :", total_ht, "| TVA :", vat_amount, "| Total TTC :", total_ttc)
````

### Exercice de synthèse ⭐⭐⭐ – Durée d'analyse d'un fichier de logs

Un script d'analyse traite **25 000 lignes par seconde**. Le fichier de logs du serveur web contient **1 250 000 lignes**, et on prévoit que le volume augmente de **40 %** le mois prochain (variable `growth_percent`).

Calculez :
1. `duration_s` : la durée d'analyse actuelle, en secondes (nombre entier, utilisez `//`) ;
2. `future_lines` : le nombre de lignes le mois prochain (nombre entier : calculez l'augmentation avec `* growth_percent // 100`) ;
3. `future_duration_s` : la durée d'analyse future, en secondes (entier) ;
4. `future_minutes` et `future_seconds` : cette durée exprimée en minutes et secondes restantes.

Ajoutez **au moins deux commentaires utiles** qui expliquent vos calculs, puis affichez une phrase de résultat.

> Les séparateurs `_` sont autorisés dans les nombres : `1_250_000` est plus lisible que `1250000`.

````python
lines_per_second = 25_000
current_lines = 1_250_000
growth_percent = 40

# Durée actuelle : nombre de lignes divisé par la vitesse de traitement
duration_s = current_lines // lines_per_second

# Volume futur : on ajoute 40 % en restant sur des nombres entiers
future_lines = current_lines + current_lines * growth_percent // 100
future_duration_s = future_lines // lines_per_second

# Conversion en minutes et secondes restantes
future_minutes = future_duration_s // 60
future_seconds = future_duration_s % 60

print("Durée actuelle :", duration_s, "s")
print("Durée future :", future_minutes, "min", future_seconds, "s")
````

<details><summary>💡 Indice</summary>

- Augmentation = `current_lines * growth_percent // 100` ; nombre futur = actuel + augmentation.
- Minutes : `//` par 60. Secondes restantes : `%` par 60.
</details>

````python
# ✅ Vérification
assert duration_s == 50
assert future_lines == 1_750_000
assert future_duration_s == 70
assert future_minutes == 1 and future_seconds == 10
print("✅ Exercice de synthèse réussi")
````
