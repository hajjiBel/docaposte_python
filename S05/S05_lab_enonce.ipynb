### Exemple guidé : fonctions de supervision

```python
def disk_usage(used_gb: float, total_gb: float) -> float:
    """Renvoie le taux d'occupation d'un disque, en pourcentage arrondi à 0,1."""
    return round(used_gb / total_gb * 100, 1)


def health(usage_percent: float, warning: float = 75, critical: float = 90) -> str:
    """Renvoie le niveau d'alerte correspondant à un taux d'occupation."""
    if usage_percent >= critical:
        return "CRITICAL"
    if usage_percent >= warning:
        return "WARNING"
    return "OK"


usage = disk_usage(389, 512)
print(usage, health(usage), health(usage, warning=80), health(usage, critical=70))
```
